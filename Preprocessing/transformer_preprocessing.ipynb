```python
import pandas as pd
import re

# LOAD DATASET

df = pd.read_csv("stress_dataset_final_merged.csv")
```

```python
# ENCODE LABELS

label_map = {
    "low": 0,
    "medium": 1,
    "high": 2
}

df["label_encoded"] = df["label"].map(label_map)
```

```python
# LIGHT CLEANING FOR TRANSFORMERS

def light_clean(text):

    text = str(text).lower()

    # remove urls
    text = re.sub(r'http\S+|www\S+', '', text)

    # remove extra spaces/newlines
    text = re.sub(r'\s+', ' ', text).strip()

    return text

df["text"] = df["text"].apply(light_clean)
```

```python
# KEEP NEEDED COLUMNS

df = df[["text", "label_encoded"]]

# REMOVE NULLS

df.dropna(inplace=True)

# REMOVE EMPTY TEXTS

df = df[df["text"].str.strip() != ""]

# REMOVE DUPLICATES

df.drop_duplicates(subset=["text"], inplace=True)
```

```python
# RESET INDEX

df.reset_index(drop=True, inplace=True)

# SAVE DATASET

df.to_csv("df_transformer_ready.csv", index=False)
```
